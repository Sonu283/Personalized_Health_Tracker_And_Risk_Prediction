```python
import pandas as pd
import numpy as np
import glob
import joblib
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.signal import find_peaks
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier, StackingClassifier, VotingClassifier
from imblearn.over_sampling import ADASYN
import xgboost as xgb
import lightgbm as lgb
import catboost as cb

# ---------------------- Load & Preprocess Data ----------------------
output_df = pd.read_csv("\ecg_dataset\output.csv")
features, labels = [], []
patient_files = glob.glob("\ecg_dataset\patient*.csv")

for file in patient_files:
    patient_id = file.split('\\')[-1].split('.')[0]
    patient_data = pd.read_csv(file)

    # Feature engineering
    stats = patient_data.describe().loc[['mean', 'std', 'min', 'max', '50%', '25%', '75%']].values.flatten()
    fft_features = np.abs(np.fft.fft(patient_data.values))[:10].mean(axis=1)
    energy = np.sum(patient_data.values**2, axis=0)
    peaks, _ = find_peaks(patient_data.values.flatten())
    peak_features = [len(peaks), np.mean(patient_data.values.flatten()[peaks]) if len(peaks) > 0 else 0]
    zero_crossings = np.sum(np.diff(np.signbit(patient_data.values)))
    skewness = patient_data.skew().values
    kurtosis = patient_data.kurtosis().values

    combined_features = np.concatenate([stats, fft_features, energy, peak_features, [zero_crossings], skewness, kurtosis])
    features.append(combined_features)

    label = output_df[output_df['Name'] == patient_id].drop(columns=['Name']).values.flatten()
    labels.append(label)

X = np.array(features)
y = np.argmax(np.array(labels), axis=1)

print("\nOriginal class distribution:")
for i in np.unique(y):
    print(f"Class {i}: {np.sum(y == i)} samples")

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

adasyn = ADASYN(random_state=42)
X_resampled, y_resampled = adasyn.fit_resample(X_scaled, y)

print("\nResampled class distribution:")
for i in np.unique(y_resampled):
    print(f"Class {i}: {np.sum(y_resampled == i)} samples")

X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.15, random_state=42, stratify=y_resampled)

# ---------------------- Model Definitions ----------------------
mlp_model = MLPClassifier(hidden_layer_sizes=(256,128,64), activation='relu', solver='adam',
                          learning_rate='adaptive', max_iter=300, early_stopping=True, random_state=42)
gb_model = GradientBoostingClassifier(n_estimators=400, learning_rate=0.01, max_depth=6, random_state=42)
xgb_model = xgb.XGBClassifier(n_estimators=500, learning_rate=0.01, max_depth=8, random_state=42)
lgb_model = lgb.LGBMClassifier(n_estimators=500, learning_rate=0.01, max_depth=10, random_state=42)
catboost_model = cb.CatBoostClassifier(iterations=400, learning_rate=0.01, depth=8, verbose=0, random_state=42)
rf_model = RandomForestClassifier(n_estimators=400, max_depth=12, random_state=42)

# ---------------------- Training All Models ----------------------
models = {
    'MLP': mlp_model, 'GB': gb_model, 'XGB': xgb_model,
    'LGB': lgb_model, 'CatBoost': catboost_model, 'RF': rf_model
}
accuracies = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    accuracies[name] = acc
    print(f"\n🔸 {name} Accuracy: {acc * 100:.2f}%")
    print(classification_report(y_test, preds))

# ---------------------- Stacking Model ----------------------
stacking_model = StackingClassifier(
    estimators=[(name.lower(), model) for name, model in models.items()],
    final_estimator=xgb.XGBClassifier(n_estimators=200, learning_rate=0.01, max_depth=3, random_state=42),
    passthrough=True
)
stacking_model.fit(X_train, y_train)
stack_preds = stacking_model.predict(X_test)
stack_acc = accuracy_score(y_test, stack_preds)
print(f"\n🔸 Stacking Accuracy: {stack_acc * 100:.2f}%")

# ---------------------- Voting Ensemble ----------------------
voting_model = VotingClassifier(
    estimators=[(name.lower(), model) for name, model in models.items()],
    voting='soft'
)
voting_model.fit(X_train, y_train)
voting_preds = voting_model.predict(X_test)
voting_acc = accuracy_score(y_test, voting_preds)
print(f"\n🔥 Voting Accuracy: {voting_acc * 100:.2f}% 🚀")
print(classification_report(y_test, voting_preds))

# ---------------------- Manual Weighted Ensemble ----------------------
probas = {name: model.predict_proba(X_test) for name, model in models.items()}
total_acc = sum(accuracies.values())
weights = {k: v / total_acc for k, v in accuracies.items()}
print(f"\nModel Weights: {weights}")

weighted_proba = sum(weights[name] * probas[name] for name in models)
final_preds = np.argmax(weighted_proba, axis=1)
weighted_acc = accuracy_score(y_test, final_preds)
print(f"\n🔥 Weighted Ensemble Accuracy: {weighted_acc * 100:.2f}% 🚀")
print(classification_report(y_test, final_preds))

# ---------------------- Confusion Matrix ----------------------
plt.figure(figsize=(10, 8))
cm = confusion_matrix(y_test, final_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - Weighted Ensemble")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()

# ---------------------- Save Models ----------------------
for name, model in models.items():
    joblib.dump(model, f"{name.lower()}_model.pkl")
joblib.dump(stacking_model, "stacking_model.pkl")
joblib.dump(voting_model, "voting_model.pkl")
joblib.dump(weights, "ensemble_weights.pkl")
joblib.dump(scaler, "scaler.pkl")

# ---------------------- Prediction Function ----------------------
def predict_ecg(file_path):
    user_data = pd.read_csv(file_path)
    stats = user_data.describe().loc[['mean', 'std', 'min', 'max', '50%', '25%', '75%']].values.flatten()
    fft_features = np.abs(np.fft.fft(user_data.values))[:10].mean(axis=1)
    energy = np.sum(user_data.values**2, axis=0)
    peaks, _ = find_peaks(user_data.values.flatten())
    peak_features = [len(peaks), np.mean(user_data.values.flatten()[peaks]) if len(peaks) > 0 else 0]
    zero_crossings = np.sum(np.diff(np.signbit(user_data.values)))
    skewness = user_data.skew().values
    kurtosis = user_data.kurtosis().values
    combined_features = np.concatenate([stats, fft_features, energy, peak_features, [zero_crossings], skewness, kurtosis])

    scaler = joblib.load("scaler.pkl")
    input_scaled = scaler.transform([combined_features])

    loaded_models = {
        'mlp': joblib.load("mlp_model.pkl"),
        'gb': joblib.load("gb_model.pkl"),
        'xgb': joblib.load("xgb_model.pkl"),
        'lgb': joblib.load("lgb_model.pkl"),
        'catboost': joblib.load("catboost_model.pkl"),
        'rf': joblib.load("rf_model.pkl")
    }

    weights = joblib.load("ensemble_weights.pkl")
    probs = {name: model.predict_proba(input_scaled) for name, model in loaded_models.items()}
    final_proba = sum(weights[name.upper()] * probs[name] for name in probs)
    final_prediction = np.argmax(final_proba, axis=1)[0]

    print("\n🔹 ECG Condition Prediction Results:")
    for name, model in loaded_models.items():
        pred = model.predict(input_scaled)[0]
        conf = np.max(model.predict_proba(input_scaled)) * 100
        print(f"{name.upper():<8}: Class {pred} (Confidence: {conf:.2f}%)")
    print(f"🏆 Final Weighted Prediction: Class {final_prediction} (Confidence: {np.max(final_proba)*100:.2f}%)")

    return final_prediction

# Example use (uncomment to test):
# predict_ecg("path_to_ecg.csv")
```
