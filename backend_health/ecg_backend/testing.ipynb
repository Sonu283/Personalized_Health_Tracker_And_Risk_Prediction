```python
import os
import joblib
import json
import pandas as pd
import numpy as np
from sklearn.metrics import classification_report, accuracy_score

# ======================== CONFIGURATION ========================
MODEL_DIR = os.path.join(os.getcwd(), "saved_model")

CONFIG = {
    "model_path": os.path.join(MODEL_DIR, "disease_prediction_model.joblib"),
    "vectorizer_path": os.path.join(MODEL_DIR, "symptom_vectorizer.joblib"),
    "label_encoder_path": os.path.join(MODEL_DIR, "label_encoder.joblib"),
    "meta_path": os.path.join(MODEL_DIR, "model_metadata.json")
}
```

```python

# ======================== LOAD COMPONENTS ========================
print("🔁 Loading saved components...")
model = joblib.load(CONFIG["model_path"])
vectorizer = joblib.load(CONFIG["vectorizer_path"])
label_encoder = joblib.load(CONFIG["label_encoder_path"])

# Load metadata if needed
with open(CONFIG["meta_path"], 'r') as f:
    metadata = json.load(f)
print(f"✅ Model: {metadata['model']} | Accuracy (stored): {metadata['accuracy']}%")

```

```python
# ======================== MANUAL SYMPTOM INPUT ========================
# Example manual input (you can change this to take dynamic user input too)
input_symptoms = ["itching", "skin rash", "nodal skin eruptions", "dischromic patches"]

# Clean symptoms the same way as training
cleaned_input = [s.strip().lower().replace(" ", "_") for s in input_symptoms]
input_text = ' '.join(cleaned_input)

# ======================== VECTORIZE & PREDICT ========================
X_manual = vectorizer.transform([input_text])
probs = model.predict_proba(X_manual)[0]
top_preds = np.argsort(probs)[::-1][:5]

print("\n🧠 Top Predictions for manual input:")
for i in top_preds:
    disease_name = label_encoder.inverse_transform([i])[0]
    print(f"👉 {disease_name} - {round(probs[i]*100, 2)}%")

```
